# Building a Frame Wing Model in SDynPy

```python
# import sdynpy as sdpy
import sdynpy.demo.frame_wing as frame_wing
geometry = frame_wing.geometry
system = frame_wing.system
system.assign_modal_damping(0.05)
modes = system.eigensolution(maximum_frequency = 1500)
modal_system = modes.system()
for i in range(6):
    modal_system.K[i,i] = 160
    modal_system.C[i,i] = 1.2
nodes = geometry.node
elements = geometry.element
mode_shapes = modal_system.eigensolution(maximum_frequency=1500, return_shape=True)
```

```python
plot_kwargs = {'node_size':3,'line_width':2,'view_from':[-1,1,1]}
geometry.plot(label_nodes=True, **plot_kwargs)
```
